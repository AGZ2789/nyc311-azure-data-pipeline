# NYC 311 Gold Analysis

## Load Gold Datasets

```python
df_complaints = spark.read.format("delta").load(
    "abfss://gold@stnyc311.dfs.core.windows.net/complaint_counts/"
)
```

```python
df_boroughs = spark.read.format("delta").load(
    "abfss://gold@stnyc311.dfs.core.windows.net/borough_counts/"
)
```

```python
df_resolution = spark.read.format("delta").load(
    "abfss://gold@stnyc311.dfs.core.windows.net/resolution_by_complaint/"
)
```

### Top 5 Complaint Types

```python
df_complaints.show(5, truncate=False)
```

### Requests by Borough

```python
df_boroughs.show(truncate=False)
```

### Median Resolution Times

```python
from pyspark.sql import functions as F

df_resolution_display = (
    df_resolution
    .withColumn(
        "median_resolution_hours",
        F.round("median_resolution_hours", 2)
    )
)

df_resolution_display.show(10, truncate=False)
```
